My Research question: How do restaurant inspection results differ across the five boroughs of New York City?

Objective: This project analyzes NYC restaurant inspection data to compare insoection scores, grades, and violations across the 5 boroughs

Analysis 1: Average Inspection Score by Borough


```python
borough_scores = df.groupby("BORO")["SCORE"].mean()
```

```python
import matplotlib.pyplot as plt

borough_scores = df.groupby("BORO")["SCORE"].mean()
borough_scores.plot(kind='bar', title='Average Inspection Score by Borough')
plt.xlabel('Borough')
plt.ylabel('Average Score')
plt.show()
```

Queens had the highest average Inspection score, meaning restaurants in Queens had more violation points on average. Staten Island had the lowest average Inspection score, meaning restaurants there had fewer violation points on average.

Analysis 2: Restaurant Grades by Borough

```python
df.groupby("BORO")["GRADE"].value_counts()
```

These raw counts aren't fair by borough comparison. What we really want to know is What percentageo of graded inspections in each borough received an A, B, or C?

N = Not Yet Graded
Z = Grade Pending
P = Grade Pending after reopening that resulted in closure

A = 0-13
B = 14-27
C = 28+

```python
abc_grades = df[df["GRADE"].isin(["A", "B", "C"])]

abc_grades.groupby("BORO")["GRADE"].value_counts()
```

```python
grade_totals = abc_grades.groupby("BORO")["GRADE"].count()
```

```python
grade_totals
```

```python
a_grades = abc_grades[abc_grades["GRADE"] == "A"].groupby("BORO")["GRADE"].count()
```

```python
a_grades
```

```python
a_percent = (a_grades / grade_totals) * 100
a_percent
a_percent.round(2)
```

```python
a_percent.round(2).plot(kind='bar', title='Percentage of A Grades by Borough')
plt.xlabel('Borough')
plt.ylabel('Percentage')
plt.show()
```

Staten Island had the highest percentage of a grades at 85.56%, while Queens had the lowest at 70.65%. This shows that the distribution of restaurant grades differ across to five boroughs.

Analysis 3: Violations by Borough

```python
df.groupby("BORO")["VIOLATION CODE"].count()
```

```python
df.groupby("BORO")["VIOLATION CODE"].count().plot(kind='bar', title='Violations by Borough')
plt.xlabel('Borough')
plt.ylabel('Number of Violations')
plt.show()
```

Manhattan had the highest number of recorded violations with 6891, while Staten Island had the lowest was 635. However, these are raw accounts, so differences in the number of restaurant inspection records in each borough may affect the totals.

CONCLUSION

The analysis shows that restaurant inspection results differ across the five boroughs of New York City. Queens had the highest average inspection score, while Staten Island had the lowest. Since lower inspection scores represent fewer violation points, Staten Island had fewer violation points on average. Staten Island also had the highest percentage of A grades at 85.56%, while Queens had the lowest at 70.65%. Manhattan had the highest total number of recorded violations, while Staten Island had the lowest; however, these totals may be affected by differences in the number of inspection records in each borough. Overall, the results show clear differences in inspection scores, grades, and recorded violations across the five boroughs.
