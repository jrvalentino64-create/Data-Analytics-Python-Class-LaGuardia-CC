```python
df['salary'].mean() # here was see the mean of the salary
```

```python
df['salary'].median() # here was see the median of the salary
```

```python
df.groupby('rank')[['salary']].mean() # here was see the mean of the salary & rank by using the groupby function
```

```python
df.groupby('rank')[['salary']].median() # here was see the median of the salary & rank by using the groupby function
```

```python
df.groupby('sex')[['salary']].mean() # here was see the mean of the salary by sex (m/f) by using the groupby function
```

```python
df.groupby('sex')[['salary']].median() # here was see the median of the salary by sex (m/f) by using the groupby function
```

```python
df = df.rename(columns={'Unnamed: 0':'Professor ID'})
# we renamed our unnamaed column to Professort ID
```
