```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Schools =['James Madison','Brooklyn Average']
scores = [66.1, 66.6]
```

```python
plt.bar(Schools, scores)
plt.title('Common Core Algebra Regents Mean Scores Comparison')
plt.ylabel('Scores')
plt.text(0, 66.1, '66.1')
plt.text(1, 66.6, '66.6')
plt.show()
```

```python
# Based on my analysis, James Madison high school's average Common Core Algebra Regent Score was .05 points lower compared with the Brooklyn average
```

My research question was: How does James Madison High School's average Common Core Algebra Regent score compare with the average score of other schools in Brooklyn?
